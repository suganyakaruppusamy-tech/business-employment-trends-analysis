##Phase 1: Statistical Analysis

```python
#Measure of Central Tendency

mean_value = df_clean["Data_value"].mean()
median_value = df_clean["Data_value"].median()
mode_value = df_clean["Data_value"].mode()[0]
print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)

```

```python
# Distribution of Employment Values
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df_clean,
    x="Data_value",
    bins=30,
    color="steelblue",
    edgecolor="black",
    alpha=0.85
)

# Mean
plt.axvline(
    x=mean_value,
    color="red",
    linestyle="--",
    linewidth=2.5,
    label=f"Mean : {mean_value:,.2f}"
)

# Median
plt.axvline(
    x=median_value,
    color="green",
    linestyle="-.",
    linewidth=2.5,
    label=f"Median : {median_value:,.2f}"
)

# Mode
plt.axvline(
    x=mode_value,
    color="darkorange",
    linestyle=":",
    linewidth=2.8,
    label=f"Mode : {mode_value:,.2f}"
)

# Title
plt.title(
    "Distribution of Employment Values with Central Tendency Measures",
    fontsize=15,
    fontweight="bold",
    pad=15
)

# Axis labels
plt.xlabel(
    "Employment Value",
    fontsize=12,
    fontweight="bold"
)

plt.ylabel(
    "Frequency",
    fontsize=12,
    fontweight="bold"
)

# Tick size
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)

# Grid
plt.grid(axis="y", linestyle="--", alpha=0.35)

# Legend
plt.legend(
    title="Statistics",
    fontsize=10,
    title_fontsize=11,
    loc="upper right",
    frameon=True
)

# Remove unnecessary borders
sns.despine()

plt.tight_layout()
plt.show()
```

##Central Tendency Interpretation
- The mean employment value (74,018) is significantly higher than the median (14,265) and mode (287), indicating that a few extremely large employment values are increasing the overall average.
- The pattern Mean > Median > Mode confirms that the employment data is positively (right) skewed, showing that employment is not evenly distributed across industries and categories.
- The median provides a more reliable measure of a typical employment value because it is less affected by extreme employment figures than the mean.
- The mode suggests that lower employment values occur most frequently, indicating that many industries or categories have relatively smaller workforces compared to a few dominant sectors.
- From a business perspective, employment is concentrated in a limited number of high-employment industries, while the majority of industries operate with comparatively smaller workforce sizes. Therefore, industry-level analysis is essential for making accurate workforce and policy decisions.

```python
# Count outliers using IQR method

Q1 = df_clean["Data_value"].quantile(0.25)
Q3 = df_clean["Data_value"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df_clean[
    (df_clean["Data_value"] < lower_limit) |
    (df_clean["Data_value"] > upper_limit)
]

print("Number of outliers:", len(outliers))
```

- A large number of observations appear above the upper whisker as outliers, showing that a few industries have exceptionally high employment levels compared to the majority of the dataset.
- The median is positioned closer to the lower quartile, while the upper whisker extends much farther than the lower whisker, confirming that the data is positively skewed.
- These outliers represent genuine business observations rather than data errors, as they correspond to large industries with naturally higher employment levels. Therefore, they should be retained for meaningful business analysis instead of being removed.
- From a business perspective, the box plot indicates that employment is concentrated in a limited number of high-employment sectors, while most industries maintain comparatively smaller workforces.
```python
# Year vs Total Filled Jobs

# Aggregate total filled jobs by year
yearly_jobs = (
    df_clean.groupby("Year", as_index=False)["Data_value"]
           .sum()
)

yearly_jobs
```

##Bivariate Analysis
**Visualization 2:Top 10 Industries by Total Filled Jobs**

Objective 2: Compare employment levels across industries.


##Purpose of the Analysis

This visualization compares employment trends over time for the industries with the highest workforce participation.

It helps answer:

- Which industries have consistently high employment?
- Do all major industries grow at the same rate?
- Are there industries with faster or slower employment growth?

```python
#Step 1: Filter the data
df_industry = df_clean[
    (df_clean["Group"] == "Industry by employment variable") &
    (df_clean["Employment_Variable"] == "Filled jobs") &
    (df_clean["Category"] != "Total Industry")&
    (df_clean["Year"] < 2026)
]
```

```python
#Step 2: Calculate total filled jobs by industry
industry_jobs = (
    df_industry
    .groupby("Category")["Data_value"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
```

```python
# chart for total filled jobs by industry
from matplotlib.ticker import FuncFormatter
colors = sns.color_palette("Blues_r", len(industry_jobs))

plt.figure(figsize=(12,6))

ax = sns.barplot(
    x=industry_jobs.values,
    y=industry_jobs.index,
    palette=colors
)

# Data Labels
for i, value in enumerate(industry_jobs.values):
    ax.text(
        value + industry_jobs.max()*0.01,
        i,
        f"{value/1_000_000:.1f} M",
        va="center",
        fontsize=10,
        fontweight="bold"
    )

# Format x-axis in Millions
ax.xaxis.set_major_formatter(
    FuncFormatter(lambda x, pos: f"{x/1_000_000:.0f} M")
)

# Titles
plt.title(
    "Top 10 Industries by Total Filled Jobs (2011–2025)",
    fontsize=15,
    fontweight="bold",
    pad=15
)

plt.xlabel(
    "Total Filled Jobs (Millions)",
    fontsize=12,
    fontweight="bold"
)

plt.ylabel(
    "Industry",
    fontsize=12,
    fontweight="bold"
)

plt.grid(axis="x", linestyle="--", alpha=0.3)

sns.despine()

plt.tight_layout()
plt.show()
```

##Interpretation for Industries by total filled jobs
- Health Care and Social Assistance recorded the highest cumulative employment during 2011–2025, with approximately 21.8 million filled jobs, indicating its significant contribution to workforce participation over the study period.
- Manufacturing and Retail Trade ranked second and third, respectively, demonstrating that these sectors have consistently maintained a large workforce and remain major contributors to employment.
- Construction, Education and Training, and Accommodation and Food Services also generated substantial employment, reflecting sustained labour demand in infrastructure, education, and service-related industries.
- The noticeable difference between the highest and lower-ranked industries suggests that employment opportunities are concentrated in a few key sectors, while the remaining industries contribute comparatively smaller shares of the workforce.
- Overall, the analysis indicates that essential service industries and production-based sectors have been the primary drivers of employment growth between 2011 and 2025, providing valuable insights for workforce planning, policy formulation, and investment decisions.

**Visualization 3: Industry × Adjustment Type × Filled Jobs**

```python
#How do employment estimates differ across industries after seasonal adjustment?

# Prepare data
industry_adjustment = (
    df_clean[
        (df_clean["Group"] == "Industry by employment variable") &
        (df_clean["Employment_Variable"] == "Filled jobs") &
        (df_clean["Category"] != "Total Industry") &
        (df_clean["Year"] < 2026)
    ]
    .groupby(["Category", "Adjustment_Type"], as_index=False)["Data_value"]
    .sum()
)

# Top 10 industries
top10 = (
    industry_adjustment
    .groupby("Category")["Data_value"]
    .sum()
    .nlargest(10)
    .index
)

industry_adjustment = industry_adjustment[
    industry_adjustment["Category"].isin(top10)
].copy()

# Convert to Millions
industry_adjustment["Filled Jobs (M)"] = (
    industry_adjustment["Data_value"] / 1_000_000
)

# Industry order
industry_order = (
    industry_adjustment
    .groupby("Category")["Filled Jobs (M)"]
    .sum()
    .sort_values(ascending=False)
    .index
)

industry_adjustment["Category"] = pd.Categorical(
    industry_adjustment["Category"],
    categories=industry_order,
    ordered=True
)

# Adjustment type order
industry_adjustment["Adjustment_Type"] = pd.Categorical(
    industry_adjustment["Adjustment_Type"],
    categories=["Actual", "Seasonally adjusted", "Trend"],
    ordered=True
)

# Plotly grouped bar chart
fig = px.bar(
    industry_adjustment,
    x="Category",
    y="Filled Jobs (M)",
    color="Adjustment_Type",
    barmode="group",
    template="plotly_white",
    color_discrete_sequence=px.colors.sequential.Blues_r,
    title="Top 10 Industries by Adjustment Type (2011–2025)"
)

# Professional hover only
fig.update_traces(
    hovertemplate=
    "<b>%{x}</b><br>" +
    "Adjustment Type: %{fullData.name}<br>" +
    "Filled Jobs: %{y:.1f} M<extra></extra>"
)

fig.update_layout(
    title=dict(
        text="<b>Top 10 Industries by Adjustment Type (2011–2025)</b>",
        x=0.5,
        font=dict(size=22)
    ),

    width=1300,
    height=700,

    xaxis=dict(
        title="<b>Industry</b>",
        tickangle=-25
    ),

    yaxis=dict(
        title="<b>Total Filled Jobs (Millions)</b>"
    ),

    legend=dict(
        title="<b>Adjustment Type</b>",
        orientation="h",
        x=0.5,
        xanchor="center",
        y=-0.45
    ),

    hovermode="x unified"
)

fig.show()
```

##Interpretation
1. Health Care and Social Assistance recorded the highest employment across all adjustment types.
The industry consistently reported the largest number of filled jobs under Actual, Seasonally Adjusted, and Trend estimates, highlighting its significant contribution to overall employment and its sustained workforce demand throughout the study period.
2. Manufacturing, Retail Trade, and Education & Training remained among the largest employing industries.
These industries consistently recorded high employment levels across all three adjustment types, indicating stable workforce participation and their continued importance to New Zealand's economy.
3. Employment estimates were highly consistent across Actual, Seasonally Adjusted, and Trend data.
For each industry, the three adjustment types produced very similar employment totals with only minor differences. This suggests that seasonal effects had a limited impact on long-term employment estimates and that the underlying employment patterns remained stable.
4. Differences between adjustment types were relatively small.
The close alignment of the Actual, Seasonally Adjusted, and Trend values indicates that seasonal adjustment mainly smooths short-term fluctuations without substantially changing the overall employment levels across industries.
5. Industry ranking remained unchanged after seasonal adjustment.
The ordering of the top industries was consistent regardless of the adjustment type. Health Care and Social Assistance remained the largest employer, followed by Manufacturing, Retail Trade, and Education & Training, demonstrating that seasonal adjustment did not alter the overall employment structure.
